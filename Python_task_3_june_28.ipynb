Class Method (uses @classmethod and cls)

Example 1

```python
class Student:
    school = "ABC School"

    @classmethod
    def show_school(cls):
        print(cls.school)

Student.show_school()
```

Example 2

```python
class Employee:
    company = "Infosys"

    @classmethod
    def company_name(cls):
        print(cls.company)

Employee.company_name()
```

Example 3

```python
class Car:
    brand = "Toyota"

    @classmethod
    def show_brand(cls):
        print(cls.brand)

Car.show_brand()
```

Example 4

```python
class Mobile:
    country = "India"

    @classmethod
    def show_country(cls):
        print(cls.country)

Mobile.show_country()
```

Example 5

```python
class Animal:
    type = "Mammal"

    @classmethod
    def show_type(cls):
        print(cls.type)

Animal.show_type()
```

Example 6

```python
class College:
    name = "VITM"

    @classmethod
    def show_name(cls):
        print(cls.name)

College.show_name()
```

Static Method (uses @staticmethod)

Exanple 1

```python
class Math:
    @staticmethod
    def add(a, b):
        print(a + b)

Math.add(10, 20)
```

Example 2

```python
class Calculator:
    @staticmethod
    def square(n):
        print(n * n)

Calculator.square(5)
```

Example 3

```python
class Message:
    @staticmethod
    def greet():
        print("Welcome")

Message.greet()
```

Example 4

```python
class Temperature:
    @staticmethod
    def celsius_to_fahrenheit(c):
        print((c * 9/5) + 32)

Temperature.celsius_to_fahrenheit(30)
```

Example 5

```python
class Shape:
    @staticmethod
    def area_square(side):
        print(side * side)

Shape.area_square(4)
```

Example 6

```python
class Bank:
    @staticmethod
    def interest():
        print("Interest Rate: 7%")

Bank.interest()
```
